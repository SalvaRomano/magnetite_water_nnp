```python
import numpy as np
```

```python
def energies_to_configurations(dataset_file, bad_energies):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
    
    bad_energies_rounded=[]
    
    
    bad_energies_float=np.transpose(np.loadtxt(bad_energies))
    print("Lenght bad energies" , len(bad_energies_float))
   
    for ene in bad_energies_float:
        bad_energies_rounded.append(round(ene, 4))
    
    
    
    
    
    # Find the indices of the bad energy lines
    bad_energy_structures = []
    index=0
    for i, line in enumerate(lines):
        if line.startswith('energy'): 
            index += 1
            if round(float(line.split()[1]), 4 ) in bad_energies_rounded:
                #print(round(float(line.split()[1]), 4 ) )
                bad_energy_structures.append(index)
   
    
    print("Lenght bad energies founded" , len(bad_energy_structures))
    return bad_energy_structures
```

```python
def print_configurations(dataset_file, structures, good_input_data, control):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
    struct_count=0
    not_count=0   
    #print(index)
    with open(good_input_data, 'w') as new_file:
        index=0
        
        for i, line in enumerate(lines):
            if line.startswith('begin'): 
                if index in structures:
                    if control:
                        bool=True
                    else:
                        bool=False
                    struct_count+=1
                else: 
                    if control:
                        bool=False
                    else:
                        bool=True
                    not_count += 1
                index += 1
            if bool:
                new_file.write(line)
    print("old lenght:", index)    
    print(" str lenght:",struct_count )       
    if control:
        print("new lenght:",struct_count) 
    else:
        print("new lenght:",not_count) 
        
    
    return 
```

```python

# Usage energies_to_configurations
dataset_file = 'committee_sixth_data'        #name of the input.data to be cleaned
energies = 'bad_energies'                   #a column txt file of energies 
bad_structure = energies_to_configurations(dataset_file, energies)

```

```python
# Usage print configurations
   
structure_list=   bad_structure     #a column txt file of structure indices (containing zero structure)
new_input_data="committe_sixth_cleaned"                            #name of the output input.data cleaned
control= False    #set True to print the structures in structure_list, False to print all except from those
print_configurations(dataset_file, structure_list, new_input_data, control)
```
