```python
import numpy as np
```

```python
def print_configurations(dataset_file, structures, good_input_data, control):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
    struct_count=0
    not_count=0   
    #print(index)
    with open(good_input_data, 'w') as new_file:
        index=0
        
        for i, line in enumerate(lines):
            if line.startswith('begin'): 
                if index in structures:
                    if control:
                        bool=True
                    else:
                        bool=False
                    struct_count+=1
                else: 
                    if control:
                        bool=False
                    else:
                        bool=True
                    not_count += 1
                index += 1
            if bool:
                new_file.write(line)
    print("old lenght:", index)    
    print(" str lenght:",struct_count )       
    if control:
        print("new lenght:",struct_count) 
    else:
        print("new lenght:",not_count) 
        
    
    return 
```

```python
dataset_file="validation.data"           #name of the input.data to be cleaned
structure_list=np.loadtxt("union_structure")       #a column txt file of structure indices (containing zero structure)
new_input_data="new_validation.data"                            #name of the output input.data cleaned
control= False    #set True to print the structures in structure_list, False to print all except from those
print_configurations(dataset_file, structure_list, new_input_data, control)
```
