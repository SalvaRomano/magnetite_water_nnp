```python
import numpy as np
import matplotlib.pyplot as plt
import os
import re
```

```python
def atomic_force_analysis(dataset_file, bound, species):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
 
    H=re.compile(r"H")
    O=re.compile(r"O")
    Fe=re.compile(r"Fe")
    
    atom_count=0
    struct_count=0
   
    bad_list=[]
    
    if species == 'H':
        print("Working with H ")
        for i , line in enumerate(lines):

            if line.startswith('atom'):
                #print(line)
                split_line=line.split()
        
                if H.search(line):
                    #print(line)
                    if np.sqrt(float(split_line[7])**2+ float(split_line[8])**2+ float(split_line[9])**2)> bound:
                        #print(struct_count)
                        bad_list.append(struct_count)
            if line.startswith('energy'): 
            
                struct_count=struct_count+1
        
    if species == 'O':
        print("Working with O ")
        for i , line in enumerate(lines):
        
            if line.startswith('atom'):
                #print(line)
                split_line=line.split()

                if O.search(line):
                    #print(line)
                    if np.sqrt(float(split_line[7])**2+ float(split_line[8])**2+ float(split_line[9])**2)> bound:
                        #print(struct_count)
                        bad_list.append(struct_count)
          
            if line.startswith('energy'): 
           
                struct_count=struct_count+1
            
    
    if species == 'Fe': 
        print("Working with Fe ")
        for i , line in enumerate(lines):
        
            if line.startswith('atom'):
                #print(line)
                split_line=line.split()
               
                if Fe.search(line):
                    if np.sqrt(float(split_line[7])**2+ float(split_line[8])**2+ float(split_line[9])**2)> bound:
                      #print(struct_count)
                      bad_list.append(struct_count)
          
            if line.startswith('energy'): 
           
                struct_count=struct_count+1
            
       
            
        
    
    return  bad_list
```

```python
def print_configurations(dataset_file, structures, good_input_data, control):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
    struct_count=0
    not_count=0   
    #print(index)
    with open(good_input_data, 'w') as new_file:
        index=0
        
        for i, line in enumerate(lines):
            if line.startswith('begin'): 
                if index in structures:
                    if control:
                        bool=True
                    else:
                        bool=False
                    struct_count+=1
                else: 
                    if control:
                        bool=False
                    else:
                        bool=True
                    not_count += 1
                index += 1
            if bool:
                new_file.write(line)
    print("old lenght:", index)    
    print(" str lenght:",struct_count )       
    if control:
        print("new lenght:",struct_count) 
    else:
        print("new lenght:",not_count) 
        
    
    return 
```

```python
dataset_file= "./input_234.data"
bound = 8
species = 'Fe'
bad_list = atomic_force_analysis(dataset_file, bound, species)

print(np.unique((bad_list)))
```

```python

structure_list=np.unique(bad_list)     #a column txt file of structure indices (containing zero structure)
new_input_data="bad_234.data"                            #name of the output input.data cleaned
control= True  #set True to print the structures in structure_list, False to print all except from those
print_configurations(dataset_file, structure_list, new_input_data, control)
```
