```python
def first_atom_line(dataset_file):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
 
    
    first_atom_line_control=True
    struct_count=0
    first_position=[]
 
    
    for i , line in enumerate(lines):
        if first_atom_line_control:
            if line.startswith('atom'):
                print(line)
                
                split_line=line.split()
                
                first_atom_line_control=False
                first_position.append([(split_line[1]) ,(split_line[2]) , (split_line[3])] )
        if line.startswith('energy'): 
            first_atom_line_control=True
            struct_count=struct_count+1
            
    
    print(struct_count)  
        
    
    return  first_position
```

```python
def atom_lines(dataset_file):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
 
    
    struct_count=0
    position=[]
    tmp=[]
    
    for i , line in enumerate(lines):
     
        if line.startswith('atom'):
              
                
            split_line=line.split()
                
        
            tmp.append([(split_line[1]) ,(split_line[2]) , (split_line[3])] )
        if line.startswith('energy'): 
            position.append(tmp)
            tmp=[]
            struct_count=struct_count+1
            
    
    print(struct_count)  
        
    
    return  position
```

```python
def print_configurations(dataset_file, structures, good_input_data, control):
    # Read the dataset file
    with open(dataset_file, 'r') as file:
        lines = file.readlines()
    struct_count=0
    not_count=0   
    #print(index)
    with open(good_input_data, 'w') as new_file:
        index=0
        
        for i, line in enumerate(lines):
            if line.startswith('begin'): 
                if index in structures:
                    if control:
                        bool=True
                    else:
                        bool=False
                    struct_count+=1
                else: 
                    if control:
                        bool=False
                    else:
                        bool=True
                    not_count += 1
                index += 1
            if bool:
                new_file.write(line)
    print("old lenght:", index)    
    print(" str lenght:",struct_count )       
    if control:
        print("new lenght:",struct_count) 
    else:
        print("new lenght:",not_count) 
        
    
    return 
```

```python
dataset_file= "./unique.data"

first_positions = atom_lines(dataset_file)

```

```python

duplicate_list=[]

for i,position in enumerate(first_positions[:len(first_positions)-1]):
    for j, control in enumerate(first_positions[i+1:]):
        if position == control:
            duplicate_list.append([i, j+i+1])
```
