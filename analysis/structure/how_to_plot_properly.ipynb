# Single Plot

````python
import matplotlib as mpl
import matplotlib.pyplot as plt

# Set font size
mpl.rcParams.update({'font.size': 9})

# Typical Widths:
# Two column: ~ 17 cm
# Single column ~ 7 cm

# Height is up to you

# DPI (up this for more resolution !NOT THE FIGURE SIZE!):
# Testing: 100-200
# Print: 300-400
# Poster: 600

cm_to_inch = 0.393701
fig_size = (7 * cm_to_inch, 6 * cm_to_inch)
dpi = 300

# Give your subplots good names, this is a great function for keeping track of subplots
fig, ax = plt.subplot_mosaic([["test_panel"]],
                             figsize = fig_size, dpi = dpi,)

# Plotting
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) + np.exp(np.linspace(0, 2, 100))

# Think about setting the linewidth correctly
ax["test_panel"].plot(data_x, data_y, lw=1, label="Data")

# ALWAYS add labels, think about units, use latex
ax["test_panel"].set_xlabel(r"$x^{(0)}$ [nm]")
ax["test_panel"].set_ylabel(r"$f(x^{(0)})$ [nm$^2$]")

# Title useful for presentations, not so much in thesis/papers
ax["test_panel"].set_title("Test Plot")

# Number your plots A B C etc.
ax["test_panel"].text(0.05, 0.95, "A", transform=ax["test_panel"].transAxes, fontsize=12, va='top', ha='left')

# Add Minor Ticks
ax["test_panel"].xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
ax["test_panel"].yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))

# Add Legend
# For the love of god, dont plot the box
# placement easy with 'best', 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center'
ax["test_panel"].legend(frameon=False, loc="upper center")

# Arrange plots such that there is no overlap
plt.tight_layout()

# OPTIONAL Save figure
# plt.savefig("figure1_raw.png", dpi=dpi, bbox_inches='tight')
````

# Multiple plots

````python
import matplotlib as mpl
import matplotlib.pyplot as plt

# Set font size
mpl.rcParams.update({'font.size': 9})

cm_to_inch = 0.393701

# Typical Widths:
# Two column: ~ 17 cm
# Single column ~ 7 cm

# Height is up to you

# DPI (up this for more resolution !NOT THE FIGURE SIZE!):
# Testing: 100-200
# Print: 300-400
# Poster: 600

fig_size = (17 * cm_to_inch, 7 * cm_to_inch)
dpi = 300

# Give your subplots good names, this is a great function for keeping track of subplots
# USE sharex or sharey if the plots share an axis
# Two plots for a whole A4 width is actually still quite large, print it to test in doubt
fig, ax = plt.subplot_mosaic([["data_A", "data_B"]],
                             figsize = fig_size, dpi = dpi, sharex=True)

########
# Plot A
########
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) + np.exp(np.linspace(0, 2, 100))

ax["data_A"].plot(data_x, data_y, lw=1, label="Data")

# ALWAYS add labels, think about units, use latex
ax["data_A"].set_xlabel(r"$x^{(0)}$ [nm]")
ax["data_A"].set_ylabel(r"$f(x^{(0)})$ [nm$^2$]")


########
# Plot B
########
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) * np.exp(-np.linspace(0, 5, 100))

ax["data_B"].plot(data_x, data_y, lw=1, label="Data")

# ALWAYS add labels, think about units, use latex
ax["data_B"].set_xlabel(r"$x^{(0)}$ [nm]")
ax["data_B"].set_ylabel(r"$g(x^{(0)})$ [nm$^2$]")

# Title useful for presentations, not so much in thesis/papers
ax["data_A"].set_title("Test Plot A")
ax["data_B"].set_title("Test Plot B")

# Number your plots A B C etc.
for i, k in enumerate(ax.keys()):
    ax[k].text(0.05, 0.95, chr(i+65), transform=ax[k].transAxes, fontsize=12, va='top', ha='left')

# Add Minor Ticks
for k in ax.keys():
    ax[k].xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
    ax[k].yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))

# Add Legend
# For the love of god, dont plot the box
# placement easy with 'best', 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center'
for k in ax.keys():
    ax[k].legend(frameon=False, loc="upper center")

# Arrange plots such that there is no overlap
plt.tight_layout()

# OPTIONAL Save figure
# plt.savefig("figure1_raw.png", dpi=dpi, bbox_inches='tight')
````

````python
import matplotlib as mpl
import matplotlib.pyplot as plt

# Set font size
mpl.rcParams.update({'font.size': 9})

cm_to_inch = 0.393701

# Typical Widths:
# Two column: ~ 17 cm
# Single column ~ 7 cm

# Height is up to you

# DPI (up this for more resolution !NOT THE FIGURE SIZE!):
# Testing: 100-200
# Print: 300-400
# Poster: 600

fig_size = (17 * cm_to_inch, 14 * cm_to_inch)
dpi = 300

# Give your subplots good names, this is a great function for keeping track of subplots
# USE sharex or sharey if the plots share an axis
fig, ax = plt.subplot_mosaic([["data_A", "data_B"],
                              ["data_C", "data_D"]],
                             figsize = fig_size, dpi = dpi, sharex=True)

########
# Plot A
########
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) + np.exp(np.linspace(0, 2, 100))

ax["data_A"].plot(data_x, data_y, lw=1, label="Data")

# No labels here since axis is shared
# ax["data_A"].set_xlabel(r"$x^{(0)}$ [nm]")
# ax["data_A"].set_ylabel(r"$f(x^{(0)})$ [nm$^2$]")


########
# Plot B
########
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) * np.exp(-np.linspace(0, 5, 100))

ax["data_B"].plot(data_x, data_y, lw=1, label="Data")

# No labels here since axis is shared
# ax["data_B"].set_xlabel(r"$x^{(0)}$ [nm]")
# ax["data_B"].set_ylabel(r"$g(x^{(0)})$ [nm$^2$]")


########
# Plot C
########
data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) * np.sin(np.linspace(0, 5, 100))

ax["data_C"].plot(data_x, data_y, lw=1, label="Data")

# ALWAYS add labels, think about units, use latex
ax["data_C"].set_xlabel(r"$x^{(0)}$ [nm]")
ax["data_C"].set_ylabel(r"$h(x^{(0)})$ [nm$^2$]")


########
# Plot D
########

data_x = np.linspace(0, 1, 100)
data_y = np.random.normal(size=100) - 2* np.sin(np.linspace(0, 5, 100))

ax["data_D"].plot(data_x, data_y, lw=1, label="Data")

# ALWAYS add labels, think about units, use latex
ax["data_D"].set_xlabel(r"$x^{(0)}$ [nm]")
ax["data_D"].set_ylabel(r"$h(x^{(0)})$ [nm$^2$]")


# Title useful for presentations, not so much in thesis/papers
ax["data_A"].set_title("Test Plot A")
ax["data_B"].set_title("Test Plot B")

# Number your plots A B C etc.
for i, k in enumerate(ax.keys()):
    ax[k].text(0.05, 0.95, chr(i+65), transform=ax[k].transAxes, fontsize=12, va='top', ha='left')

# Add Minor Ticks
for k in ax.keys():
    ax[k].xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
    ax[k].yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))

# Add Legend
# For the love of god, dont plot the box
# placement easy with 'best', 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center'
for k in ax.keys():
    ax[k].legend(frameon=False, loc="upper center")

# Arrange plots such that there is no overlap
plt.tight_layout()

# OPTIONAL Save figure
# plt.savefig("figure1_raw.png", dpi=dpi, bbox_inches='tight')
````
